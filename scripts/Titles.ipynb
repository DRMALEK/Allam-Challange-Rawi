```python
import re
from docx import Document
from docx.enum.text import WD_ALIGN_PARAGRAPH

def is_title(paragraph):
    text = paragraph.text.strip()
    if not text:
        return False

    # Check if the text starts with a number (including Arabic numerals)
    if not re.match(r'^[0-9\u0660-\u0669]+', text):
        return False

    # Check if the paragraph is centered
    # If alignment is None, it's typically left-aligned
    if paragraph.alignment != WD_ALIGN_PARAGRAPH.CENTER:
        return False

    # Check if the number of words is less than 6
    word_count = len(text.split())
    if word_count >= 7:
        return False

    return True

def print_titles(docx_file):
    doc = Document(docx_file)
    titles_found = False
    for paragraph in doc.paragraphs:
        if is_title(paragraph):
            print(paragraph.text)
            titles_found = True
    if not titles_found:
        print("No titles found.")

if __name__ == "__main__":
    docx_file = 'C:/Users/Sukkar/Desktop/3.docx'  # Replace with your document's filename
    print_titles(docx_file)
```
